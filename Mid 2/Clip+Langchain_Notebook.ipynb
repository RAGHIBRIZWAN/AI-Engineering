# Example of CLiP

```python
from transformers import pipeline

# Load the zero-shot image classifier (downloads CLIP automatically)
classifier = pipeline(task="zero-shot-image-classification", model="openai/clip-vit-base-patch32")
```

```python
# Run it on an image URL with custom text labels
results = classifier(
    image="https://encrypted-tbn0.gstatic.com/licensed-image?q=tbn:ANd9GcSXT0gXcft7cdFHndlUdvMZ7PQmvnSLQ1owJbiQrc_SX-HUhRBLxFW8Ae302jHo4PM4npYnGzY_NHW3hVY",
    candidate_labels=["spicy biryani with potato", "light brown pulao", "nihari"]
)

results
```

```python
from langchain_core.prompts import PromptTemplate

# Define a template with placeholders
template = """You are a helpful culinary assistant.
Explain what a {dish_name} is, including its key ingredients like {key_ingredient} and its origin.
"""

prompt_template = PromptTemplate.from_template(template)

# Format the prompt with dynamic values
formatted_prompt = prompt_template.format(
    dish_name="Biryani",
    key_ingredient="rice and spices"
)

print(formatted_prompt)
```

```python
from langchain_core.prompts import PromptTemplate

# 1. Define a complex template simulating a personalized, context-aware chatbot
system_instruction_template = """You are an expert AI {role_specialty}.
Your task is to assist {user_name}, who is {user_tier} tier member.

Context on user preferences:
- Cuisine preferences: {cuisine_preferences}
- Allergies/Restrictions: {dietary_restrictions}

Respond to their request: "{user_request}" with a customized recipe and detailed step-by-step instructions.
"""

prompt = PromptTemplate.from_template(system_instruction_template)

# 2. Simulate fetching raw, complex data from a Database/API
fictional_db_user_record = {
    "username": "Alex",
    "membership": "Platinum Premium",
    "history": {
        "cuisines": ["Indian", "Mexican", "Thai"],
        "dietary": ["Gluten-Free", "Peanuts-Free"]
    }
}

# 3. Simulate a dynamic runtime user input
current_user_request = "I want to make a quick dinner with chicken."

# 4. Programmatically bind and pre-format the database structures into the template
# This highlights LangChain's utility in translating complex app states into strings cleanly
formatted_complex_prompt = prompt.format(
    role_specialty="Culinary Chef and Nutritionist",
    user_name=fictional_db_user_record["username"],
    user_tier=fictional_db_user_record["membership"],
    cuisine_preferences=", ".join(fictional_db_user_record["history"]["cuisines"]),
    dietary_restrictions=", ".join(fictional_db_user_record["history"]["dietary"]),
    user_request=current_user_request
)

print(formatted_complex_prompt)
```
